```python
# unique professions

profession_data = []

professions = data['Profession'].unique() # give all the professions
# print(professions)
for profession in professions:
    total = len(data[data['Profession'] == profession])
    defaults = len(data[(data['Profession'] == profession) & (data['Risk_Flag'] == 1)])
    rate = (defaults / total) * 100
    profession_data.append((profession, rate))

    # print(f"{profession}: Total = {total}, Defaulters = {defaults}, Default Rate = {rate:.2f}%")

# Sort and print top 5
profession_data.sort(key=lambda x: x[1], reverse=True)

for prof in profession_data[:5]:
    print(f"{prof[0]}: Default Rate = {prof[1]:.2f}%")
```

```python
# Filter defaulters in each range
less_than_2 = data[(data['CURRENT_JOB_YRS'] <= 2) & (data['Risk_Flag'] == 1)]
two_to_five = data[(data['CURRENT_JOB_YRS'] > 2) & (data['CURRENT_JOB_YRS'] <= 5) & (data['Risk_Flag'] == 1)]
more_than_five = data[(data['CURRENT_JOB_YRS'] > 5) & (data['Risk_Flag'] == 1)]

# Total records in each job experience range
total_less_than_2 = data[(data['CURRENT_JOB_YRS'] <= 2)]
total_two_to_five = data[(data['CURRENT_JOB_YRS'] > 2) & (data['CURRENT_JOB_YRS'] <= 5)]
total_more_than_five = data[(data['CURRENT_JOB_YRS'] > 5)]

# Print totals
print(f"Total defaulters with ≤2 years in job: {len(less_than_2)}")
print(f"Total defaulters with 2–5 years in job: {len(two_to_five)}")
print(f"Total defaulters with >5 years in job: {len(more_than_five)}")

# Calculate probabilities
prob_less_than_2 = (len(less_than_2) / len(total_less_than_2)) * 100
prob_two_to_five = (len(two_to_five) / len(total_two_to_five)) * 100
prob_more_than_five = (len(more_than_five) / len(total_more_than_five)) * 100

# Print results
print(f"Default probability (≤2 years): {prob_less_than_2:.4f}%")
print(f"Default probability (2–5 years): {prob_two_to_five:.4f}%")
print(f"Default probability (>5 years): {prob_more_than_five:.4f}%")
```
